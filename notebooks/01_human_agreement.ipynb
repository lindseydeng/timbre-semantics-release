# Experiment 1: agreement with human timbre ratings

Start by viewing saved results. Fresh inference is an optional last step.
Install the project in this kernel environment with `pip install -e ".[notebooks]"`.

The experiment builds two aligned matrices, each **61 instruments × 16 descriptors**:
- `human_ratings`: mean listener ratings.
- `similarities`: cosine similarity of audio embeddings to descriptor embeddings.

A column-wise Pearson correlation answers: does this descriptor rank instruments like humans do?
A row-wise correlation answers: does this instrument have the same descriptor profile?


```python
from pathlib import Path
import pandas as pd

# Run Jupyter from the repository root or from notebooks/.
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
if not (ROOT / "pyproject.toml").exists():
    raise FileNotFoundError("Open this notebook from the repository root or notebooks/.")
```

## 1. Inspect saved descriptor correlations

These are already computed results, not a new model run.


```python
descriptor_results = pd.read_csv(ROOT / "results/reference/experiment1/descriptors.csv")
display(descriptor_results.pivot(index="descriptor", columns="model", values="pearson"))
```

## 2. Inspect saved instrument profiles


```python
MODEL = "muq-mulan"
chinese = pd.read_csv(ROOT / f"results/reference/experiment1/{MODEL}_chinese.csv")
western = pd.read_csv(ROOT / f"results/reference/experiment1/{MODEL}_western.csv")
print("Chinese mean Pearson r:", chinese.pearson.mean())
print("Western mean Pearson r:", western.pearson.mean())
```
